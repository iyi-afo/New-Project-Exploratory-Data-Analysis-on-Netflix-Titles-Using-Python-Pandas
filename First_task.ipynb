```python
import pandas as pd
```

```python
netflix = pd.read_csv(r"C:\Users\DELL\Desktop\Data_Engineering Dataset\netflix_titles.csv")
netflix.head()
```

```python
netflix.isnull().sum()
```

```python
netflix["country"]= netflix["country"].fillna("Unknown")
netflix.head()
```

```python
netflix.shape
```

```python
netflix= netflix.dropna(subset = "date_added")
```

```python
netflix.isnull().sum()
```

```python
netflix["type"].value_counts()
```

```python
netflix["date_added"] = pd.to_datetime(netflix["date_added"].str.strip(), format = "mixed")
```

```python
netflix.head()
```

```python
country_split = netflix["country"].str.split(",")
country_split
```

```python
country_explode = country_split.explode()
country_explode
```

```python
Top_10_countries = country_explode.value_counts()
```

```python
Top_10_countries
```

```python
netflix["year_added"] = netflix["date_added"].dt.year
netflix.head()
```

```python
old_shows= netflix[netflix["release_year"] < 2000]
old_shows
```

```python
old_shows.value_counts().sum()
```

```python
listed_in_split = netflix["listed_in"].str.split(",")
listed_in_split
```

```python
listed_in_explode = listed_in_split.explode()
listed_in_explode
```

```python
Most_common_genre = listed_in_explode.value_counts()
Most_common_genre.head(10)
```
